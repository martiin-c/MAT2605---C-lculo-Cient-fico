### ¿Qué es Jupyter Notebook?
Es un entorno de trabajo interactivo que permite intercalar bloques de texto (como este, escrito en Markdown), ecuaciones, visualizaciones y bloques de código ejecutable. Anaconda ya incluye por defecto Jupyter Notebook y las librerías de computación científica (NumPy, SciPy, Matplotlib, Pandas), por lo que estamos listos para comenzar.

### 1. Variables y Tipos de Datos
En Python, no es necesario declarar el tipo de variable explícitamente. El lenguaje lo infiere dinámicamente en el momento de la asignación.

```python
# Variables y tipos de datos elementales en Python
entero = 10              # int (Integer)
decimal = 3.14159        # float (Punto flotante)
texto = "Hola MAT2605"   # str (String o cadena de texto)
booleano = True          # bool (Booleano: True o False)

# Imprimir variables y sus tipos
print(f"El valor es {decimal} y su tipo es {type(decimal)}")
print(f"El valor es {texto} y su tipo es {type(texto)}")
```

### 2. Control de Flujo (If, Elif, Else)
Las sentencias condicionales permiten ejecutar distintos bloques de código dependiendo de si se cumple o no una condición. En Python, la indentación (los espacios al inicio de la línea) define los bloques de código.

```python
# Control de flujo básico
valor = 5

if valor > 0:
    print("El valor es estrictamente positivo")
elif valor == 0:
    print("El valor es exactamente cero")
else:
    print("El valor es negativo")
```

### 3. Ciclos (For loops)
Los ciclos `for` en Python se utilizan para iterar sobre una secuencia (como una lista, una tupla o un rango de números).

```python
# Iterando sobre una lista
mi_lista = [2, 4, 6, 8]
print("Elementos de la lista:")
for numero in mi_lista:
    print(numero)

print("\nIterando con la función range():")
# range(inicio, fin, paso) - el 'fin' no se incluye
for i in range(0, 5): 
    print(f"Iteración número {i}")
```

### 4. Definición de Funciones
Las funciones permiten empaquetar y reutilizar código. Se definen utilizando la palabra reservada `def`.

```python
# Creación de una función matemática simple
def evaluar_signo(numero):
    """Esta función recibe un número y retorna una cadena de texto indicando su signo."""
    if numero > 0:
        return "Positivo"
    elif numero < 0:
        return "Negativo"
    else:
        return "Cero"

# Llamamos a la función
resultado = evaluar_signo(-12.5)
print("El número ingresado es:", resultado)

# Lambda function
funlambda = lambda x: x**2+x -1
funlambda(0.1)
```

### 5. Computación Científica: NumPy y Matplotlib
Python base no está optimizado para operaciones matemáticas intensivas. Para el cálculo científico y análisis numérico, utilizamos **NumPy** (operaciones vectorizadas y matriciales).

```python
# Importamos las librerías
import numpy as np
import matplotlib.pyplot as plt

# --- Diferencia entre Listas y Arrays de NumPy ---
lista_a = [1, 2, 3]
lista_b = [4, 5, 6]
# La suma de listas las concatena
print("Suma de listas de Python:", lista_a + lista_b) 

array_a = np.array([1, 2, 3])
array_b = np.array([4, 5, 6])
# La suma de arrays realiza la operación elemento por elemento (mucho más eficiente)
print("Suma de arrays de NumPy:", array_a + array_b)

# 2d arrays: Crear matrices
A = np.array([[1, 2, 3], [4, 5, 6]])  # matriz 2x3, primera fila 1 2 3, segunda fila 4 5 6
B = np.zeros((3, 3))                  # matrix 3x3, todos los coeficientes = 0

# acceso a coeficientes (Cuidado: Enumeracion 0,1,2,...)
print("Coeficiente primera fila, segunda columna de matriz A:", A[0,1])
print("Tercer coeficiente del array array_a:", array_a[2])

# Producto euclideano entre dos vectores del mismo tamaño:
print("Producto interior:", np.inner(array_a, array_b))
                   # se evalua: 1*4 + 2*5 + 3*6

# matrix vector product
print("Producto matriz-vector:",np.matmul(A,array_a))
# alternativa con @ (a partir de Python 3.5)
print("Producto matriz-vector con operador @:",A@array_a)

# Resolver un sistema lineal: x1+x2 = 1, 3*x1+5*x2=2, escribir en forma Ax = b
A = np.array([[1, 2], [3, 5]])
b = np.array([1, 2])

# Resolver sistema lineal Ax = b
x = np.linalg.solve(A, b)
print("Solucion del sistema lineal:", x)
```

### 6. Visualización de Datos **Matplotlib** (visualización de datos).
Utilizaremos `matplotlib.pyplot` para generar gráficos de manera rápida.

```python
# Crear un vector de valores espaciados uniformemente
# np.linspace(inicio, fin, número de puntos)
x = np.linspace(-10, 10, 100)

# Aplicar funciones matemáticas vectorizadas
y1 = np.sin(x)
y2 = np.cos(x)

# Generar el gráfico
plt.figure(figsize=(8, 4)) # Tamaño de la figura
plt.plot(x, y1, label="sen(x)", color="blue", linewidth=2)
plt.plot(x, y2, label="cos(x)", color="red", linestyle="--")

plt.title("Gráfico de funciones trigonométricas")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.legend() # Mostrar las etiquetas (labels)

# Mostrar el gráfico
plt.show()
```

### Pregunta 2: Funciones y Visualización
Crea una función en Python llamada `f(x, alpha)` que evalúe la expresión:
$$f(x) = x^{2\alpha}$$

Luego, genera un vector `x` con valores entre 0 y 5 (usa 100 puntos). Utiliza un ciclo `for` para graficar la función en una sola figura para los siguientes valores de $\alpha$: 1, 0.5, 0.25 y 0.125. Asegúrate de incluir una leyenda (legend) que indique qué curva corresponde a cada $\alpha$.

<div class="alert alert-block alert-success"><b> Soluci&oacuten: </b>
</div>

```python
# tu codigo aqui
def f(x,alpha):
    exponente = 2*alpha
    return x**exponente

x = np.linspace(0, 5, 100) #vector
#funciones dependientes de alpha
a1 = f(x,1)
a2 = f(x,0.5)
a3 = f(x,0.25)
a4 = f(x,0.125)

# Generar el gráfico
plt.figure(figsize=(8, 4)) # Tamaño de la figura
plt.plot(x, a1, label="f con a = 1", color="blue", linewidth=2)
plt.plot(x, a2, label="f con a = 0.5", color="red", linestyle="--")
plt.plot(x, a3, label="f con a = 0.25", color="black", linewidth=5)
plt.plot(x, a4, label="f con a = 0.125", color="pink", linestyle="--")

plt.title("Gráfico de funciones dependientes de alpha")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.legend() # Mostrar las etiquetas (labels)

# Mostrar el gráfico
plt.show()
```

### Pregunta 4: Gráfica de una Función Física
La altura $h(t)$ de un objeto lanzado hacia arriba está dada por:
$$h(t) = v_0 t - \frac{1}{2} g t^2$$
Donde $v_0 = 20$ m/s y $g = 9.8$ m/s$^2$.

Crea un vector de tiempo $t$ utilizando NumPy que vaya desde $t=0$ hasta $t=4$ segundos con 50 puntos. Calcula el vector de alturas $h$ y **crea un gráfico de $h$ en función de $t$**. 
Asegúrate de agregar título al gráfico, etiquetas en los ejes X e Y, y activar la grilla (`grid`).
<div class="alert alert-block alert-success"><b> Soluci&oacuten: </b>
</div>

```python
# tu codigo aqui
def altura(v,t,g):
    primer_termino = v*t
    segundo_termino = (g*t**2)/2
    return primer_termino - segundo_termino

tiempo = np.linspace(0, 4, 50) #vector
#funciones dependientes de alpha
caso1 = altura(20,tiempo,9.8)

# Generar el gráfico
plt.figure(figsize=(8, 4)) # Tamaño de la figura
plt.plot(tiempo, caso1, label="Altura con v0 = 20 m/s ", color="blue", linewidth=2)


plt.title("Gráfico de altura")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.legend() # Mostrar las etiquetas (labels)

# Mostrar el gráfico
plt.show()
```
